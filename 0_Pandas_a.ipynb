# Euro 12

## Import data from datasets/euro12.csv

```python
FILEPATH = "/content/drive/MyDrive/MLinAPP_Data/euro12.csv"
euro12 = pd.read_csv(FILEPATH)
```

##  Calculate the mean Yellow Cards given per Team

```python
mean_yellow_cards = euro12['Yellow Cards'].mean()
print(f"The mean Yellow Cards given per Team is {mean_yellow_cards}")
```

## Which teams scored more than 6 goals?

```python
high_scorers = euro12[euro12['Goals'] > 6]
print("teams with higher goal>6 :", high_scorers['Team'])
```

## Select the teams that start with G

```python
teams_g = euro12[euro12['Team'].str.startswith('G')]
print("teams that start with G :", teams_g['Team'])
```

## Present only the Shooting Accuracy from England, Italy and Russia

```python
accuracy_eir = euro12.loc[euro12['Team'].isin(['England', 'Italy', 'Russia']), ['Team', 'Shooting Accuracy']]
print(accuracy_eir)
```

# Drinks

## Read and save data from datasets/drinks.txt

```python
FILEPATH = "/content/drive/MyDrive/MLinAPP_Data/drinks.csv"
drinks = pd.read_csv(FILEPATH, sep=',')
```

## Which continent drinks more beer on average?

```python
top_beer_continent = drinks.groupby('continent')['beer_servings'].mean().idxmax()
print(f"The continent that drinks more beer on average is {top_beer_continent}")
```

## For each continent print the statistics for wine consumption.

```python
wine_stats = drinks.groupby('continent')['wine_servings'].describe()
print(wine_stats)
```

##  Print the mean, min and max values for spirit consumption per continent.

```python
spirit_stats = drinks.groupby('continent')['spirit_servings'].agg(['mean', 'min', 'max'])
print(spirit_stats)
```
