#### Stack arrays a and b vertically and then horizontally

```python
# Input
a = np.arange(10).reshape(2, -1)
b = np.repeat(1, 10).reshape(2, -1)

print("vertically stacked arrays:")
# Solution
vert_stack = np.vstack((a, b))
print(vert_stack)

print("horizontally stacked arrays:")
# Solution
horiz_stack = np.hstack((a, b))
print(horiz_stack)
```

#### Find positions where elements of a and b match

```python
# Input
a = np.array([1, 2, 3, 2, 3, 4, 3, 4, 5, 6])
b = np.array([7, 2, 10, 2, 7, 4, 9, 4, 9, 8])

# Solution
matches = np.where(a == b)
print("Matching indices:", matches)
print("Matching indices:", matches[0])
```

####  Get all items between 5 and 10 from a

#### Get the positions of top 5 maximum values in a given array a

```python
# Input
np.random.seed(100)
a = np.random.uniform(1, 50, 20)
print(a)

# Solution:
top_5_pos = np.argsort(a)[-5:][::-1]
print("Indices of the top 5 largest values:", top_5_pos)
top_5_values = a[top_5_pos]
print("Top 5 largest values:", top_5_values)
```

#### Compute the min-by-max (i.e. min/max) for each row for given 2d numpy array (see [apply_along_axis](https://numpy.org/doc/stable/reference/generated/numpy.apply_along_axis.html#numpy-apply-along-axis))

```python
# Input
np.random.seed(100)
a = np.random.randint(1,10, [5,3])
print(a)

# Solution
min_by_max = np.apply_along_axis(lambda x: np.min(x) / np.max(x), arr=a, axis=1)
print("Min by max for each row:", min_by_max)
```

####  Find the duplicate entries in the given numpy array and mark them as True. First time occurrences should be False.

```python
# Input
np.random.seed(100)
a = np.random.randint(0, 5, 10)
print('Array: ', a)

# Solution
duplicates = np.full(a.shape, True)
unique_positions = np.unique(a, return_index=True)[1]
print('Unique positions:', unique_positions)
duplicates[unique_positions] = False
print('Duplicate positions:', np.where(duplicates)[0])
```

#### Compute the moving average of window size 3, for the given 1D array
Make use of numpy.convolve

```python
# Input
arr = np.array([8, 8, 3, 7, 7, 0, 4, 2, 5, 2])

# Solution
moving_avg = np.convolve(arr, np.ones(3)/3, mode='valid')
print("Moving average with window size 3:", moving_avg)
```

#### Find the mean, median, standard deviation of iris's sepallength (1st column)
See np.genfromtxt, load only the 1st column with the appropiate data type

```python
# Solution
sepallength = np.genfromtxt(FILEPATH, delimiter=',', dtype='float', usecols=[0])
mean_sl, med_sl, std_sl = np.mean(sepallength), np.median(sepallength), np.std(sepallength)
print("Mean:", mean_sl)
print("Median:", med_sl)
print("Standard deviation:", std_sl)
```

####  Find the mean, median, standard deviation of iris's sepallength (1st column) for each unique labels

```python
# Solution
for label in np.unique(iris[:, 4]):
    mask = iris[:, 4] == label
    sl_filtered = sepallength[mask]
    print("just show 10 of recordes:", sl_filtered[:10])
    print(f"{label}: Mean={np.mean(sl_filtered):.2f}, Median={np.median(sl_filtered):.2f}, Std={np.std(sl_filtered):.2f}")
```

####  Create a new column for volume in iris, where volume is (pi x petallength x sepal_length^2)/3

```python
# Solution
# Create a new column for volume in iris, where volume is (pi x petallength x sepal_length^2)/3
sepallength = iris[:, 0].astype('float')
petallength = iris[:, 2].astype('float')
volume = (np.pi * petallength * (sepallength ** 2)) / 3
volume = volume[:, np.newaxis] # Reshape to a column vector
iris_with_volume = np.hstack([iris, volume])
iris_with_volume[:3]
```
